```python
import numpy as np
```

```python
X=np.array([
    [1,1],
    [1,2],
    [2,3],
    [2,4],
    [4,5]
    ])
```

```python
y=np.array([
    [0],
    [0],
    [1],
    [1],
    [0]
])
```

```python
X.shape
```

```python
y.shape
```

```python
np.random.seed(42)
n_features=X.shape[1]
w=np.random.randn(n_features)
b=np.random.randn(1)
```

```python
def step_function(z):
  return np.where(z>=0,1,0)
```

```python
'''
z= x.w+b
formula for weighet sum
'''
z=np.dot(X,w)+b;
pred=step_function(z)
```

```python
print(z)
print("preditcion is:-",pred)
```

```python
'''calulating the error'''
error=y-pred
print("actual value:-",y)
print("predicted value:-",pred)
print("the error in the model is:-",error)
```

```python
# ==========================================
# CELL 8 : Perceptron Learning Algorithm
# ==========================================

learning_rate = 0.1

epochs = 20

for epoch in range(epochs):

    for i in range(len(X)):

        # Forward Pass

        z = np.dot(X[i], w) + b

        pred= step_function(z)

        error = y[i] - pred

        # Weight Update Rule

        w = w + learning_rate * error * X[i]

        # Bias Update Rule

        b= b + learning_rate * error

    print(f"Epoch {epoch+1}")

    print("Weights :", w)

    print("Bias :", b)
```

```python
  z = np.dot(X, w) + b
pred = step_function(z)
print("Final Predictions")
print(pred)
print("Actual Labels")
print(y)
```

```python
accuracy = np.mean(pred == y)
```
